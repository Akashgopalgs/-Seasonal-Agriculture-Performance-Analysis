## Handle missing values

```python
# Fill missing rainfall using the median rainfall of the same season

df["Rainfall_mm"] = df.groupby("Season")["Rainfall_mm"].transform(
    lambda x: x.fillna(x.median()))

# Fill missing soil moisture using the median soil moisture of the same season

df["Soil_Moisture_pct"] = df.groupby("Season")["Soil_Moisture_pct"].transform(
    lambda x: x.fillna(x.median()))

# Fill missing yield using the median yield for the same crop and season

df["Yield_Tonnes_Ha"] = df.groupby(["Crop", "Season"])["Yield_Tonnes_Ha"].transform(
    lambda x: x.fillna(x.median()))
```

#### Create useful business-friendly measures.  

```python
# profit margin 
df["Profit_Margin_pct"] = (df["Profit_INR"] / df["Revenue_INR"]) * 100

# Cost per tonne
df['Cost_per_Tonne_INR'] = (df['Total_Cost_INR']/df['Production_Tonnes'])

# Revenue per hectare
df['Revenue_per_Hectare_INR'] = (df['Revenue_INR']/df['Farm_Area_Hectares'])
```

#### Verify financial calculations

```python
# Check whether profit equals revenue minus total cost

calculated_profit = df["Revenue_INR"] - df["Total_Cost_INR"]

difference = df["Profit_INR"] - calculated_profit

print("Maximum difference:", difference.abs().max())
```

The financial calculations are internally consistent.

### Statistical test

```python
# Separate yield values by season

kharif = df[df["Season"] == "Kharif"]["Yield_Tonnes_Ha"]
rabi = df[df["Season"] == "Rabi"]["Yield_Tonnes_Ha"]
zaid = df[df["Season"] == "Zaid"]["Yield_Tonnes_Ha"]

# Test whether the yield distributions differ between seasons

statistic, p_value = kruskal(kharif, rabi, zaid)

print("Kruskal-Wallis statistic:", round(statistic, 2))
print("p-value:", p_value)
```

```python
if p_value < 0.05:
    print("There is statistically significant evidence that yield differs across seasons.")
else:
    print("There is not enough statistical evidence to conclude that yield differs across seasons.")
```

##### Statistical Interpretation

- The Kruskal-Wallis test produces a p-value below 0.05.

- Therefore, there is statistically significant evidence that yield distributions differ across the three seasons.

- This supports the descriptive analysis showing differences in seasonal yield.

- However, statistical significance does not mean that season alone explains the differences in yield. Crop type, environmental conditions, farming practices, and other factors may also contribute to the observed variation.
